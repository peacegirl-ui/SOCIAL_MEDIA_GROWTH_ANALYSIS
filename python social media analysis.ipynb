# EDA
(EXPLOTARY DATA ANALYSIS)

```python
platform_analysis = df.groupby("Platform").agg(
    Total_Reach=("Reach", "sum"),
    Total_Engagements=("Total_Engagements", "sum"),
    Followers_Gained=("Followers_Gained", "sum"),
    Avg_Engagement_Rate=("Engagement_Rate_%", "mean")
).sort_values("Total_Engagements", ascending=False)
```

```python
import matplotlib.pyplot as plt

platform_analysis["Total_Engagements"].plot(
    kind="bar",
    figsize=(8,5)
)

plt.title("Total Engagements by Platform")
plt.xlabel("Platform")
plt.ylabel("Total Engagements")
plt.xticks(rotation=0)
plt.show()
```

```python
platform_analysis[["Avg_Engagement_Rate"]]
```

```python
platform_analysis["Avg_Engagement_Rate"].plot(
    kind="bar",
    figsize=(8,5)
)

plt.title("Average Engagement Rate by Platform")
plt.xlabel("Platform")
plt.ylabel("Engagement Rate (%)")
plt.xticks (rotation=0)



plt.show()
```

```python
content_analysis = df.groupby("Content_Type").agg(
    Total_Engagements=("Total_Engagements", "sum"),
    Total_Reach=("Reach", "sum"),
    Avg_Engagement_Rate=("Engagement_Rate_%", "mean")
).sort_values("Total_Engagements", ascending=False)

content_analysis
```

```python
ax = content_analysis["Total_Engagements"].plot(
    kind="bar",
    figsize=(9,5)
)

plt.title("Total Engagement by Content Type")
plt.xlabel("Content Type")
plt.ylabel("Total Engagements")
plt.xticks(rotation=0)

plt.show()
```

```python
content_analysis[["Avg_Engagement_Rate"]]
```

```python
reach_platform = df.groupby("Platform")["Reach"].sum().sort_values(ascending=False)
reach_platform
```

```python
country_analysis = df.groupby("Country").agg(
    Total_Engagements=("Total_Engagements", "sum"),
    Total_Reach=("Reach", "sum")
).sort_values("Total_Engagements", ascending=False)

country_analysis
```

```python
country_analysis["Total_Engagements"].plot(
    kind="barh",
    figsize=(9,5)
)

plt.title("Total Engagement by Country")
plt.xlabel("Total Engagements")
plt.ylabel("Country")
plt.show()
```

```python
age_analysis = df.groupby("Age_Group").agg(
    Total_Engagements=("Total_Engagements", "sum"),
    Total_Reach=("Reach", "sum")
).sort_values("Total_Engagements", ascending=False)

age_analysis
```

```python
age_analysis["Total_Engagements"].plot(
    kind="bar",
    figsize=(8,5)
)

plt.title("Total Engagement by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Total Engagements")
plt.xticks(rotation=0)
plt.show()
```
